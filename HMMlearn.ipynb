```python
import numpy as np

# Number of states and observations
n = int(input("Enter number of states: "))
m = int(input("Enter number of observation symbols: "))

# States
states = input("Enter state names: ").split()

# Observation symbols
symbols = input("Enter observation symbols: ").split()

# Initial probabilities
print("Enter Initial State Probabilities:")
pi = np.array(list(map(float, input().split())))

# Transition Matrix
print("Enter Transition Probability Matrix:")
A = []
for i in range(n):
    A.append(list(map(float, input().split())))
A = np.array(A)

# Emission Matrix
print("Enter Emission Probability Matrix:")
B = []
for i in range(n):
    B.append(list(map(float, input().split())))
B = np.array(B)

# Observation sequence
obs = input("Enter Observation Sequence: ").split()

# Convert observations into indices
obs_index = [symbols.index(x) for x in obs]

# Forward Algorithm
alpha = np.zeros((len(obs), n))

# Initialization
for i in range(n):
    alpha[0][i] = pi[i] * B[i][obs_index[0]]

# Recursion
for t in range(1, len(obs)):
    for j in range(n):
        alpha[t][j] = np.sum(alpha[t-1] * A[:, j]) * B[j][obs_index[t]]

# Termination
prob = np.sum(alpha[-1])

print("\nForward Probability Table:")
print(alpha)

print("\nLikelihood of Observation Sequence =", prob)
```
