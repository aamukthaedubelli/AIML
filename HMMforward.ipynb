```python
import numpy as np
from hmmlearn import hmm

# Number of states and observation symbols
n = int(input("Enter number of states: "))
m = int(input("Enter number of observation symbols: "))

# Initial probabilities
print("Enter Initial State Probabilities:")
pi = np.array(list(map(float, input().split())))

# Transition matrix
print("Enter Transition Probability Matrix:")
A = []
for i in range(n):
    A.append(list(map(float, input().split())))
A = np.array(A)

# Emission matrix
print("Enter Emission Probability Matrix:")
B = []
for i in range(n):
    B.append(list(map(float, input().split())))
B = np.array(B)

# Create HMM
model = hmm.CategoricalHMM(n_components=n)
model.startprob_ = pi
model.transmat_ = A
model.emissionprob_ = B

# Observation sequence
print("Observation Encoding:")
print("0 = Low, 1 = Medium, 2 = High")

obs = list(map(int, input("Enter observation sequence: ").split()))
obs = np.array(obs).reshape(-1, 1)

# Likelihood
log_prob = model.score(obs)

print("\nLikelihood =", np.exp(log_prob))

# Additional sequences
print("\nEnter number of additional sequences:")
k = int(input())

for i in range(k):
    seq = list(map(int, input(f"Sequence {i+1}: ").split()))
    seq = np.array(seq).reshape(-1, 1)
    print(f"O{i+1} =", np.exp(model.score(seq)))

print("\nForward Algorithm and hmmlearn produce the same likelihood.")
print("hmmlearn internally uses the Forward Algorithm.")
```
